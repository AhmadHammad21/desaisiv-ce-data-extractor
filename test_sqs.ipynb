```python
import boto3
import json
import uuid

# Initialize the SQS client
sqs = boto3.client('sqs')

# Your FIFO Queue URL
queue_url = "https://sqs.us-east-1.amazonaws.com/983376079397/ce-data-extractor-dev.fifo"

data = {
    "bucket": "broker-claim-exp-report-files-dev",
    "file_key": "123/1000/Data.zip"
}

try:
    response = sqs.send_message(
        QueueUrl=queue_url,
        MessageBody=json.dumps(data),
        MessageGroupId="group1",
        MessageDeduplicationId=str(uuid.uuid4())  # Safer deduplication
    )

    print("Message Sent! Message ID:", response['MessageId'])

except Exception as e:
    print(f"Failed to send message: {str(e)}")
```
