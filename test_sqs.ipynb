```python
import boto3

# Initialize the SQS client
sqs = boto3.client('sqs')

# Your FIFO Queue URL
queue_url = "https://sqs.us-east-1.amazonaws.com/983376079397/ce-data-extractor-dev.fifo"

# Send a message to the FIFO queue
response = sqs.send_message(
    QueueUrl=queue_url,
    MessageBody="This is a test message",
    MessageGroupId="group1",  # Ensure this is used for FIFO processing
    MessageDeduplicationId="unique-id-123"  # Optional, but important for deduplication
)

print("Message Sent! Message ID:", response['MessageId'])
```
